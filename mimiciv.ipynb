# Constructing dataframe for department destination assist

```python
# Constructing df for department destination identification

query = """
SELECT ad.hadm_id,
       ad.subject_id,
       ser.prev_service,
       ser.curr_service,
       ser.transfertime,
       tran.transfer_id,
       tran.eventtype,
       tran.careunit,
       tran.intime,
       tran.outtime,
       ad.hospital_expire_flag
FROM mimiciv_hosp.admissions ad
JOIN mimiciv_hosp.services ser
ON ad.hadm_id = ser.hadm_id
JOIN mimiciv_hosp.transfers tran
ON ad.hadm_id = tran.hadm_id
WHERE tran.eventtype = 'transfer'
ORDER BY ad.hadm_id, tran.intime
"""

cur.execute(query)
rows = cur.fetchall()
df_services = pd.DataFrame(rows, columns=[desc[0] for desc in cur.description])
df_services.head(10)
```

```python
# Printing the number of records in df_services
print(f"Number of records in df_services: {len(df_services)}")
```

```python
# Gathering distinct careunits from df_services
distinct_careunits = df_services['careunit'].unique()

# Creating dictionary with distinct care units as keys and None as default values
careunit_dict = {careunit: None for careunit in distinct_careunits}

# Matching MIMIC-IV careunit names with Keck department and level of care names
careunit_dict[distinct_careunits[0]] = ('Vascular Surgery', 'ICU')
careunit_dict[distinct_careunits[16]] = ('Vascular Surgery', 'MedSurg')
careunit_dict[distinct_careunits[13]] = ('Cardiology - General', 'ICU')
careunit_dict[distinct_careunits[8]] = ('Cardiology - General', 'MedSurg')
careunit_dict[distinct_careunits[19]] = ('Cardiology - General', 'Telemetry')
careunit_dict[distinct_careunits[1]] = ('Cardiac Surgery', 'MedSurg')
careunit_dict[distinct_careunits[12]] = ('Neurology', 'MedSurg')
careunit_dict[distinct_careunits[20]] = ('Neurology', 'Stepdown')
careunit_dict[distinct_careunits[21]] = ('Neurology', 'ICU')
careunit_dict[distinct_careunits[23]] = ('Neurology', 'Telemetry')
careunit_dict[distinct_careunits[14]] = ('Hematology (Med CC)', 'Stepdown')
careunit_dict[distinct_careunits[15]] = ('Hematology (Med CC)', 'Telemetry')
careunit_dict[distinct_careunits[17]] = ('GYN', 'MedSurg')



# Printing careunit_dict in tabulated format
print("Care Unit Information:")
for careunit, services in careunit_dict.items():
    print(f"{careunit}: {services}")
```

```python
# Dropping keys from careunit_dict that have None as value (careunits that have no Keck match)
careunit_dict = {k: v for k, v in careunit_dict.items() if v is not None}

# Filtering out records from df_services that have None value in for careunit in careunit_dict
df_filtered = df_services[df_services['careunit'].isin(careunit_dict.keys())]
print(f"Filtered DataFrame shape: {df_filtered.shape[0]} rows, {df_filtered.shape[1]} columns")
```

```python
# Replacing careunit names with Keck department names and adding a column for level of care
df_filtered['department'] = df_filtered['careunit'].map(lambda x: careunit_dict[x][0])
df_filtered['level_of_care'] = df_filtered['careunit'].map(lambda x: careunit_dict[x][1])
df_filtered.head()
```

```python
# Randomly assigning requested level of care to each record
# Level of care must only increase or decrease by one level in the order of ICU-Stepdown-Telemetry-MedSurg
# Valid requested levels based on the current level
adjacent_levels = {
    "ICU": ["Stepdown"],
    "Stepdown": ["ICU", "Telemetry"],
    "Telemetry": ["Stepdown", "MedSurg"],
    "MedSurg": ["Telemetry"]
}

rng = np.random.default_rng(seed=42)

df_filtered["requested_level_of_care"] = df_filtered["level_of_care"].map(
    lambda level: rng.choice(adjacent_levels[level])
)
df_filtered.head()
```
