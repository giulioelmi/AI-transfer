```python
import pandas as pd
import psycopg2
import os

# Connect to the PostgreSQL database
conn = psycopg2.connect(
    host="localhost",
    database="mimiciv_demo",
    user="postgres",
    password=os.getenv("DB_PWORD")
)

# Create a cursor object to execute SQL queries
cur = conn.cursor()

# Example query to fetch data from a table
cur.execute("SELECT * FROM mimiciv_hosp.patients")
rows = cur.fetchall()
df = pd.DataFrame(rows, columns=[desc[0] for desc in cur.description])
df.head()
```

```python
# Fetch and display the first few rows from mimiciv_hosp.transfers
cur.execute("SELECT * FROM mimiciv_hosp.transfers")
rows = cur.fetchall()
df_transfers = pd.DataFrame(rows, columns=[desc[0] for desc in cur.description])
df_transfers.head()
```
